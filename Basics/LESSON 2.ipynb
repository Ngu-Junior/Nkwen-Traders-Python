##  Stock Status Logic

```python
rice_50kg =[0,1,10,11,20]
```

```python
for quantity in rice_50kg:
    if quantity == 0:
        print("Out of Stock : ",quantity)
    elif 0<quantity<11:
        print("Low Stock : ",quantity)
    else:
        print("Well Stocked : ",quantity)
```
