```python
products = [
    {
        "name": "Rice",
        "price": 34000,
        "quantity": 10
    },
    {
        "name": "Oil",
        "price": 1400,
        "quantity": 8
    },
    {
        "name": "Sugar",
        "price": 750,
        "quantity": 15
    },
    {
        "name": "Milk",
        "price": 2500,
        "quantity": 12
    },
    {
        "name": "Bread",
        "price": 1000,
        "quantity": 20
    }
]
```

## Each dictionary represents one row (one product), while each key represents a column such as name, price, or quantity.

## - looping throuhg a dictionary

```python
for product in products:
    print(f"Product: {product['name']} | Quantity: {product['quantity']} | Price: {product['price']}")
```

## Calculating Total Inventory Value

```python
total_inventory_value = 0
for product in products:
    total_inventory_value = total_inventory_value + (product["price"] * product["quantity"])

print(f"Total Inventory Value: {total_inventory_value}")
```
