### Stock Classification Function

```python
def classify_stock(quantity):
    if quantity == 0:
        return "Out of Stock"
    elif quantity <= 10:
        return "Low Stock"
    else:
        return "Well Stocked"
```

```python
new_quantities = [0,1,10,11,20]

for quantity in new_quantities:
    print(f"Quantity: {quantity} : {classify_stock(quantity)}")
```

## Inventory Value Function

```python
def calculate_inventory_value(products):
    total = 0

    for product in products:
        total = total + (product["price"] * product["quantity"])

    return total
```

```python
new_products = [
    {"name": "Soap", "price": 1500, "quantity": 12},
    {"name": "Biscuits", "price": 1000, "quantity": 25},
    {"name": "Matches", "price": 2500, "quantity": 6},
    {"name": "Cassava", "price": 4000, "quantity": 9},
    {"name": "Tomatoes", "price": 800, "quantity": 30}
]
total = calculate_inventory_value(new_products)

print(f"Total Inventory Value: {total}")
```
