# Lists, Tuples & Loops

## LISTS

```python
product = ["Rice","Beans","Bread","Oil","Maggi"]
```

```python
#indexing
print(product[0])
#print(product[5]) :list index out of range
print(product[4])
print(product[-1])
print(product[:])
```

```python
#list slicing
print(product[0:3])
print(product[-4:-2])
```

```python
# changing a product
product[1] = "Salt"
```

```python
print(product)
```

## Loops

```python
numbers = [10,34,54,9,20]
```

```python
sum = 0
```

```python
for x in numbers:
    sum += x
print(sum)
average = sum/len(numbers)
print(average)
```

```python
# summing the kwen dataset
total = 0
for x in daily_sales:
    total += x
print(total)
average1 = total/len(numbers)
print(average1)
```

```python
#Single Highest Sales
max = daily_sales[0]
for x in daily_sales:
    if x > max:
        max = x
single_highest_sales = max
print(single_highest_sales )
```

```python
print(total)
print(average1)
print(single_highest_sales)
```
