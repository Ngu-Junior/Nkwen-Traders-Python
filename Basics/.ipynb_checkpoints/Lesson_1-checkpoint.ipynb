# One Kwen Traders Product

```python
product_name = "Rice 50Kg"
```

```python
quantity = 4
```

```python
price = 37414.0
```

```python
print(product_name, type(product_name))
print(quantity, type(quantity))
print(price, type(price))
```

```python
total_stock_value = price * quantity

print(total_stock_value)
```

```python
print(f"Nkwen Traders has {quantity} units of {product_name} in stock at {price:.2f} FCFA each, giving a total stock value of {total_stock_value:.2f} FCFA.")
```
